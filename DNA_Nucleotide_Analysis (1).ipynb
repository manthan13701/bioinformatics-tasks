This notebook counts each nucleotide in a DNA sequence, calculates its GC content, and plots the nucleotide counts as a bar chart.

## Step 1: Define the DNA sequence

```python
dna = 'ATGCGATACGCTTACGATCGATCGCTAGCTAGCTAGCTGATCGA'
print('Sequence:', dna)
print('Length  :', len(dna))
```

## Step 2: Count each nucleotide

```python
count_A = dna.count('A')
count_T = dna.count('T')
count_G = dna.count('G')
count_C = dna.count('C')
```

## Step 3: Print the results

```python
print(f'A: {count_A}')
print(f'T: {count_T}')
print(f'G: {count_G}')
print(f'C: {count_C}')
```

## Step 4: Calculate GC content
GC content is the percentage of bases that are guanine (G) or cytosine (C). It is an important measure in bioinformatics because it relates to DNA stability and helps in gene and genome characterization.

```python
gc = (count_G + count_C) / len(dna) * 100
print(f'GC Content: {gc:.2f}%')
```

## Step 5: Count using a dictionary

```python
counts = {base: dna.count(base) for base in 'ATGC'}
print(counts)
```

## Step 6: Plot a bar chart of the counts

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
bars = plt.bar(counts.keys(), counts.values(),
               color=['#2E86AB', '#E4572E', '#76B041', '#F2A541'])
plt.title('Nucleotide Counts in DNA Sequence')
plt.xlabel('Nucleotide')
plt.ylabel('Count')
for bar in bars:
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.1,
             int(bar.get_height()), ha='center')
plt.tight_layout()
plt.show()
```

The notebook counted each nucleotide, calculated the GC content, and visualized the base composition of the sequence.
