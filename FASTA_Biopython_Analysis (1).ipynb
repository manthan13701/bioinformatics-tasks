## Step 2: Import the required modules

```python
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction
```

## Step 3: Upload the FASTA file
Download the file from NCBI first: go to ncbi.nlm.nih.gov, search **insulin human**, open a result, click **Send to > Complete Record > File > FASTA**, and save it. To get several sequences in one file, tick multiple results on the search page before using Send to.

Then run the next cell and choose the downloaded file.

```python
try:
    from google.colab import files
    uploaded = files.upload()
    fasta_file = list(uploaded.keys())[0]
except ImportError:
    # Running outside Colab: put the file name here
    fasta_file = 'sequence.fasta'

print('Using file:', fasta_file)
```

## Step 4: Parse the FASTA file

```python
records = list(SeqIO.parse(fasta_file, 'fasta'))
print(f'Number of sequences found: {len(records)}')
```

## Step 5 and 6: ID, length, GC content and first 50 bases of each record

```python
for r in records:
    print(f'ID: {r.id}')
    print(f'Length: {len(r.seq)}')
    print(f'GC: {gc_fraction(r.seq)*100:.2f}%')
    print(f'Sequence: {str(r.seq[:50])}...')
    print('-' * 60)
```

## Step 7: Summary table of all sequences

```python
import pandas as pd

summary = pd.DataFrame({
    'ID': [r.id for r in records],
    'Length (bp)': [len(r.seq) for r in records],
    'GC (%)': [round(gc_fraction(r.seq) * 100, 2) for r in records],
})
summary
```

## Conclusion
Biopython's `SeqIO` parsed the FASTA file, and `gc_fraction` gave the GC content of each sequence. The summary table lets us compare sequence IDs, lengths and GC content at a glance.
